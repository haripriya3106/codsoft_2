```python
import pandas as pd

# ==========================================
# 1. IMPORT THE DATASET
# ==========================================

df = pd.read_csv("/kaggle/input/datasets/suvidyasonawane/eda-practice-dataset/EDA_Practice_Dataset.csv")

print("Dataset imported successfully!")

# ==========================================
# 2. INSPECT THE DATASET
# ==========================================

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nNumber of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe(include="all"))


# ==========================================
# 3. IDENTIFY MISSING VALUES
# ==========================================

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())


# ==========================================
# 4. IDENTIFY DUPLICATE RECORDS
# ==========================================

print("\nNumber of duplicate records:")
print(df.duplicated().sum())


# ==========================================
# 5. IDENTIFY INCONSISTENT DATA
# ==========================================

print("\nUnique values in categorical columns:")

for column in df.select_dtypes(include="object").columns:
    print("\n", column)
    print(df[column].unique())


# ==========================================
# 6. HANDLE MISSING VALUES
# ==========================================

# Numeric columns → fill missing values with median
numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())


# Categorical columns → fill missing values with mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])


# ==========================================
# 7. REMOVE DUPLICATE RECORDS
# ==========================================

df = df.drop_duplicates()

print("\nDuplicates after cleaning:")
print(df.duplicated().sum())


# ==========================================
# 8. REMOVE EXTRA SPACES
# ==========================================

for column in categorical_columns:
    df[column] = df[column].str.strip()


# ==========================================
# 9. CORRECT DATA TYPES
# ==========================================

print("\nData types before correction:")
print(df.dtypes)

# Convert date columns if your dataset has a Date column
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Convert Age column if present
if "Age" in df.columns:
    df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

print("\nData types after correction:")
print(df.dtypes)


# ==========================================
# 10. FINAL CHECK
# ==========================================

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate records after cleaning:")
print(df.duplicated().sum())

print("\nCleaned dataset:")
print(df.head())


# ==========================================
# 11. SAVE CLEANED DATASET
# ==========================================

df.to_csv("cleaned_dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
print("File name: cleaned_dataset.csv")
```
